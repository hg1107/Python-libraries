```python
import numpy as np
```

```python
grid = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(grid[1, 2]) # Single element
print(grid[:, 1]) # Column
print(grid[1, :]) # Row
print(grid[0, : : -1]) # Reverse row
```
