## Arrays

- It converts every elements to a single datatype

```python
arr = np.array([1, 2, 3, 4, 5, 6])
arr
```

```python
arr = np.array([1, 3.14, "Hello"])
arr
```

```python
arr = np.array([1, 2, 3, 4, 5.1])
arr
```

- Multiple dimensions
    - 2D arrays
    - 3D arrays
    - etc.

```python
two_d = np.array([
    [1, 2, 3],
    [4, 5, 6]])

two_d
```

```python
two_d.shape     # Two rows, three columns
```

### Creating arrays:

```python
np.zeros((3, 3))
```

```python
np.ones((3, 3))
```

```python
# For a specific number:
np.full((3, 3), 5)
```

```python
np.empty((3, 3))
```

```python
np.eye(3)
```

```python
np.diag([1, 2, 3])
```

```python
np.arange(0, 10, 2)   # Start, stop, step
```

```python
np.linspace(0, 10, 5)       # Five evenly spaced numbers between 0 and 10
```

```python
np.logspace(0, 10, 5)       # Five logarithmically spaced numbers between 0 and 10
```

#### Random arrays

```python
seed = np.random.seed((42))
rand = np.random.rand( 3)
randn = np.random.randn(3)
randint = np.random.randint(1, 10, size = 3)
uniform = np.random.uniform(1, 10, size = 3)
choice = np.random.choice([1, 2, 3, 4, 5])

print(seed, "\n")
print(rand, "\n")
print(randn, "\n")
print(randint, "\n")
print(uniform, "\n")
print(choice, "\n")
```
