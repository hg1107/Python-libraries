```python
import numpy as np

arr = np.random.randint(0, 10, size = (3, 6))
arr
```

```python
arr = arr.reshape(6, 3)       # The multiple of reshaped array size should be same as that of og array
arr
```

```python
arr.transpose()             # Transpose of matrix
```

```python
arr.flatten()       # Convert matrix in 1D, always returns a copy
```

```python
arr.ravel()         # Convert matrix in 1D, a view when possible, faster
```

```python
arr.swapaxes(0, 1)          # Swap any two axes, useful beyond 2D
```
