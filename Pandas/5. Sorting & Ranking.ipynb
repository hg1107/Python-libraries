```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('employees_100.csv')
df.head()
```

## Sorting:

```python
df.sort_values(by = "salary")
```

```python
df.sort_values(by = 'salary', ascending=False)
```

```python
df['department'] = df['department'].str.strip().str.lower()
```

```python
# Department A - Z, salary high-to-low:
df.sort_values(
    by = ['department', 'salary'],
    ascending = [True, False]
)
```

## Ranking

```python
df.drop_duplicates(subset = ['employee_id'], keep = "first", inplace=True)
```

```python
df['salary'] = df['salary'].str.strip().str.replace(",", "")

df['salary_rank'] = df['salary'].rank(ascending = False)
df
```

```python
df.sort_values(by = 'salary_rank')
```

## Top and Bottom Records

```python
df['salary'] = df['salary'].str.replace(',', '', regex = False)
df['salary'] = pd.to_numeric(df['salary'], errors= 'coerce')
df['salary'] = df['salary'].fillna(df['salary'].median(), inplace = True)
```

```python
df.nlargest(5, 'salary')
```

```python
df.nsmallest(5, 'performance_score')
```
