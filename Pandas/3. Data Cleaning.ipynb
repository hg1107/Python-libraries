```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("employees_100.csv")
df.head()
```

### Finding missing values

```python
df.isnull().sum()
```

```python
# Dropping rows with null values
df2 = df.dropna()
df2
```

```python
df2.isnull().sum()
```

```python
# Dropping null values for a specific column
df2 = df.dropna(subset = ['city'])
df2
```

```python
df.head()
```

## Filling null values

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df['salary'] = df['salary'].str.replace(',', '', regex = False)
df['salary'] = pd.to_numeric(df['salary'], errors= 'coerce')
df['salary'] = df['salary'].fillna(df['salary'].median())
```

```python
df.isnull().sum()
```

```python
df['department'].value_counts()
```

```python
df['department'] = df['department'].fillna(df['department'].mode()[0])
```

```python
df['city'] = df['city'].fillna(df['city'].mode()[0])
```

```python
df.isnull().sum()
```

## Duplicates

```python
df.duplicated().sum()
```

```python
df = df.reset_index()
df.duplicated(subset = ['employee_id']).sum()
```

```python
df.drop_duplicates(subset = ['employee_id'], keep = "first")
```

## Renaming Columns

```python
df.rename(columns = {"emp_name": "employee_name"}, inplace = True)
```

```python
# Cleaning every column names:
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```
