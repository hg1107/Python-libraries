### df.loc works if an indexed column is present, and returns value based on the value present in the indexed column

```python
df.loc[[184, 145]]
```

```python
df.loc[184:145]
```

### df.iloc works by giving the index position of the dataframe, like a string index

```python
df.iloc[0 : 10]
```

```python
df.loc[154 : 145][['name', 'city', 'age', 'salary']]    # [rows][cols]
```

## Boolean Filtering

```python
df['age'] > 40
```

```python
df[df['age'] > 40]
```

## Logical

```python
# People working in Finance and age greater than 30
df[(df['department'] == "Finance") & (df['age'] > 30) & (df['experience'] > 5)]
```

## Filtering Methods

```python
df[df['department'].isin(['Engineering', "Date Science"])]
```

```python
df[df["experience"].between(3, 5)]
```

```python
df[df['name'].str.contains("Verma")]
```
