```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('employees_100.csv')
df.head()
```

## Creating new Column:

```python
df['salary'] = df['salary'].str.replace(',', '', regex = False)
df['salary'] = pd.to_numeric(df['salary'], errors= 'coerce')
df['salary'] = df['salary'].fillna(df['salary'].median(), inplace = True)
```

```python
df['annual_salary'] = df['salary'] * 12
```

```python
df.head()
```

## Apply() function

```python
def categorize(num):
    if num > 5:
        return "Experienced Candidate"
    elif num <= 5:
        return "Learner"

df['experience_status'] = df['experience'].apply(categorize)
df.head()
```

## Map Function

```python
department_codes = {
    "engineering": "ENG",
    "marketing": "MKT",
    "finance": "FIN",
    "data science": "DAS"
}

df["department_codes"] = df['department'].str.lower().map(department_codes)
df.head()
```
